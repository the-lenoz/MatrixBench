# MatrixBench: benchmark report

This notebook analyses core timings and the CLI end-to-end series separately. Run `matrixbench_bench` and `scripts/run_cli_benchmarks.py` first, then set `RESULTS_DIR` below. Gauss (`double`) and Bareiss (`long long`) are separate experiments; they are not a direct algorithm-versus-algorithm comparison.

```python
from __future__ import annotations
import csv, json, os, platform
from collections import defaultdict
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np

RESULTS_DIR = Path(os.environ.get('MATRIXBENCH_RESULTS_DIR', '../benchmark-results/quick'))
RNG = np.random.default_rng(20261023)
assert RESULTS_DIR.is_dir(), f'Missing results directory: {RESULTS_DIR.resolve()}'
with (RESULTS_DIR / 'metadata.json').open(encoding='utf-8') as source:
    CORE_METADATA = json.load(source)
print('Core run:', json.dumps(CORE_METADATA, indent=2))
print('Notebook platform:', platform.platform())
print('Notebook processor:', platform.processor() or 'unavailable')
```

```python
def read_csv(name):
    with (RESULTS_DIR / name).open(newline='', encoding='utf-8') as source:
        return list(csv.DictReader(source))

core = read_csv('core_samples.csv')
validation = read_csv('validation.csv')
cli = read_csv('cli_samples.csv') if (RESULTS_DIR / 'cli_samples.csv').exists() else []
for row in core + cli:
    row['size'] = int(row['size']); row['time_ns'] = float(row['time_ns'])
print(f'{len(core):,} core samples, {len(validation):,} validation rows, {len(cli):,} CLI samples')

def grouped(rows, *columns):
    result = defaultdict(list)
    for row in rows:
        result[tuple(row[column] for column in columns)].append(row)
    return result

def median_ci(values, iterations=2000):
    values = np.asarray(values, dtype=float)
    median = np.median(values)
    if len(values) < 2:
        return median, median, median
    samples = RNG.choice(values, size=(iterations, len(values)), replace=True)
    bounds = np.quantile(np.median(samples, axis=1), [0.025, 0.975])
    return median, *bounds

success_core = [row for row in core if row['status'] == 'success']
```

```python
# Median runtime with bootstrap 95% CI, faceted by algorithm and input family.
determinants = [r for r in success_core if r['phase'] == 'determinant']
families = sorted({r['family'] for r in determinants})
algorithms = ('gauss', 'bareiss')
fig, axes = plt.subplots(2, len(families), figsize=(4 * len(families), 7), sharex=True)
for ai, algorithm in enumerate(algorithms):
    for fi, family in enumerate(families):
        ax = axes[ai, fi]
        subset = [r for r in determinants if r['algorithm'] == algorithm and r['family'] == family and r['magnitude'] == '1']
        for storage in ('contiguous', 'jagged'):
            by_size = grouped([r for r in subset if r['storage'] == storage], 'size')
            xs = sorted(int(size[0]) for size in by_size)
            points = [median_ci([r['time_ns'] for r in by_size[(size,)]]) for size in xs]
            if xs:
                med = np.array([p[0] for p in points]); lo = np.array([p[1] for p in points]); hi = np.array([p[2] for p in points])
                ax.plot(xs, med, marker='o', label=storage); ax.fill_between(xs, lo, hi, alpha=.2)
        if ax.lines: ax.set_xscale('log', base=2); ax.set_yscale('log')
        ax.set_title(f'{algorithm}: {family}'); ax.grid(alpha=.25)
axes[0, 0].set_ylabel('ns / determinant'); axes[1, 0].set_ylabel('ns / determinant')
for ax in axes[-1]: ax.set_xlabel('matrix size n')
axes[0, -1].legend(); fig.suptitle('Core determinant time, median and bootstrap 95% CI'); fig.tight_layout()
```

```python
# n^3-normalized runtime and contiguous/jagged median speedup.
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
for algorithm in algorithms:
    subset = [r for r in determinants if r['algorithm'] == algorithm and r['magnitude'] == '1']
    for storage in ('contiguous', 'jagged'):
        by_size = grouped([r for r in subset if r['storage'] == storage], 'size')
        xs = sorted(int(k[0]) for k in by_size)
        ys = [np.median([r['time_ns'] for r in by_size[(n,)]]) / max(n ** 3, 1) for n in xs]
        axes[0].plot(xs, ys, marker='o', label=f'{algorithm}/{storage}')
axes[0].set_xscale('log', base=2); axes[0].set_yscale('log'); axes[0].set_xlabel('n'); axes[0].set_ylabel('ns / n³'); axes[0].legend(); axes[0].grid(alpha=.25)
speed = {}
for key, rows in grouped([r for r in determinants if r['magnitude'] == '1'], 'algorithm', 'family', 'magnitude', 'size', 'storage').items():
    speed[key] = np.median([r['time_ns'] for r in rows])
for algorithm in algorithms:
    points = []
    for key, contiguous in speed.items():
        if key[0] == algorithm and key[-1] == 'contiguous':
            jagged = speed.get(key[:-1] + ('jagged',))
            if jagged: points.append((int(key[3]), jagged / contiguous))
    if points:
        points.sort(); axes[1].plot(*zip(*points), marker='o', label=algorithm)
axes[1].axhline(1, color='black', linewidth=1); axes[1].set_xscale('log', base=2); axes[1].set_xlabel('n'); axes[1].set_ylabel('jagged / contiguous median time'); axes[1].legend(); axes[1].grid(alpha=.25)
fig.suptitle('Scaling and layout speedup'); fig.tight_layout()
```

```python
# Storage-operation costs and sample distributions for the largest available point.
storage_rows = [r for r in success_core if r['algorithm'] == 'storage']
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
for phase in ('construct', 'fill_row_major', 'copy'):
    for storage in ('contiguous', 'jagged'):
        subset = [r for r in storage_rows if r['phase'] == phase and r['storage'] == storage]
        by_size = grouped(subset, 'size')
        xs = sorted(int(k[0]) for k in by_size)
        if xs: axes[0].plot(xs, [np.median([r['time_ns'] for r in by_size[(n,)]]) for n in xs], marker='o', label=f'{phase}/{storage}')
if axes[0].lines: axes[0].set_xscale('log', base=2); axes[0].set_yscale('log')
axes[0].set_xlabel('n'); axes[0].set_ylabel('ns / operation');
if axes[0].lines: axes[0].legend(fontsize=8)
axes[0].grid(alpha=.25)
largest = max(r['size'] for r in determinants)
distribution = [r for r in determinants if r['size'] == largest and r['family'] == 'dense_pivot' and r['magnitude'] == '1']
labels = []; values = []
for key, rows in grouped(distribution, 'algorithm', 'storage').items(): labels.append('/'.join(key)); values.append([r['time_ns'] for r in rows])
if values: axes[1].boxplot(values, tick_labels=labels); axes[1].set_yscale('log'); axes[1].set_ylabel('ns / determinant')
axes[1].set_title(f'Distribution at n={largest}, dense_pivot'); fig.tight_layout()
```

```python
# Gauss error and false-singularity diagnostics. Empty/NaN references are intentionally excluded.
gauss_validation = [r for r in validation if r['algorithm'] == 'gauss' and r['status'] == 'success']
points = []
for row in gauss_validation:
    try: expected, actual = float(row['expected']), float(row['actual'])
    except ValueError: continue
    if np.isfinite(expected) and np.isfinite(actual):
        numerator = abs(actual - expected); denominator = max(abs(expected), np.finfo(float).tiny)
        error = 0.0 if numerator == 0 else (1e16 if np.log10(numerator) - np.log10(denominator) > 16 else numerator / denominator)
        points.append((int(row['size']), row['magnitude'], row['family'], error, expected != 0 and actual == 0))
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
for family in sorted({p[2] for p in points}):
    subset = [p for p in points if p[2] == family]
    axes[0].scatter([p[0] for p in subset], [max(p[3], 1e-16) for p in subset], label=family, alpha=.7)
axes[0].set_xscale('log', base=2); axes[0].set_yscale('log'); axes[0].set_xlabel('n'); axes[0].set_ylabel('relative error'); axes[0].legend(); axes[0].grid(alpha=.25)
false_by_magnitude = defaultdict(lambda: [0, 0])
for _, magnitude, _, _, false_singular in points: false_by_magnitude[magnitude][0] += int(false_singular); false_by_magnitude[magnitude][1] += 1
labels = sorted(false_by_magnitude, key=float); axes[1].bar(labels, [false_by_magnitude[x][0] / false_by_magnitude[x][1] for x in labels]); axes[1].set_xlabel('coefficient scale'); axes[1].set_ylabel('false-singular share'); axes[1].set_title('Absolute epsilon sensitivity'); fig.tight_layout()
```

```python
# Bareiss overflow frontier: fraction of timed samples that completed.
bareiss = [r for r in core if r['algorithm'] == 'bareiss' and r['phase'] == 'determinant']
ranges = sorted({int(r['magnitude']) for r in bareiss}) or [0]; dimensions = sorted({r['size'] for r in bareiss}) or [0]
heatmap = np.full((len(ranges), len(dimensions)), np.nan)
for ri, value_range in enumerate(ranges):
    for ni, size in enumerate(dimensions):
        rows = [r for r in bareiss if int(r['magnitude']) == value_range and r['size'] == size and r['family'] == 'dense_random']
        if rows: heatmap[ri, ni] = np.mean([r['status'] == 'success' for r in rows])
fig, ax = plt.subplots(figsize=(10, 4)); image = ax.imshow(heatmap, aspect='auto', vmin=0, vmax=1, cmap='viridis'); ax.set_xticks(range(len(dimensions)), dimensions); ax.set_yticks(range(len(ranges)), ranges); ax.set_xlabel('n'); ax.set_ylabel('integer coefficient range'); ax.set_title('Bareiss success fraction: dense random inputs'); fig.colorbar(image, ax=ax, label='success fraction'); fig.tight_layout()
```

```python
# The CLI is a separate experiment: process creation, parsing, I/O and output are included.
if not cli:
    print('CLI data is absent. Run scripts/run_cli_benchmarks.py to enable this section.')
else:
    fig, ax = plt.subplots(figsize=(9, 4.5))
    for algorithm in algorithms:
        for storage in ('contiguous', 'jagged'):
            subset = [r for r in cli if r['algorithm'] == algorithm and r['storage'] == storage and r['status'] == 'success']
            by_size = grouped(subset, 'size'); xs = sorted(int(k[0]) for k in by_size)
            ax.plot(xs, [np.median([r['time_ns'] for r in by_size[(n,)]]) for n in xs], marker='o', label=f'{algorithm}/{storage}')
    ax.set_xscale('log', base=2); ax.set_yscale('log'); ax.set_xlabel('n'); ax.set_ylabel('wall-clock ns / CLI invocation'); ax.grid(alpha=.25); ax.legend(); ax.set_title('CLI end-to-end time'); fig.tight_layout()
```

```python
# Compact, data-driven conclusions.
layout_wins = [value for key, value in speed.items() if key[-1] == 'contiguous' and speed.get(key[:-1] + ('jagged',), value) > value]
false_count = sum(point[-1] for point in points)
overflow_count = sum(row['status'] == 'overflow' for row in bareiss)
print(f'Contiguous storage is faster in {len(layout_wins)} comparable aggregate points (ties and magnitude-specific details remain in the plots).')
print(f'Gauss has {false_count} known-nonsingular validation cases returned as zero under the default absolute epsilon.')
known_bareiss = [r for r in validation if r['algorithm'] == 'bareiss' and r['expected'] and r['status'] == 'success']
mismatches = sum(int(r['expected']) != int(r['actual']) for r in known_bareiss)
print(f'Bareiss recorded {overflow_count} overflow timing samples; treat its dense high-range region as outside long long representability.')
print(f'Bareiss known-answer mismatches: {mismatches} of {len(known_bareiss)} successful validation cases.')
print('Interpret all absolute timings only within this machine/compiler/run; compare medians and confidence intervals, not single samples.')
```
